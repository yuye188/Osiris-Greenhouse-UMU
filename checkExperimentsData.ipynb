```python
import pandas as pd
```

```python
# read multiple CSV files and combine them into a single DataFrame
def readListCSV(csv_files):
    df = pd.DataFrame()
    for file in csv_files:
        temp_df = pd.read_csv(file)
        temp_df['_time'] = pd.to_datetime(temp_df['_time'], format='mixed', utc=True)
        # remove the timezone information from the _time column
        temp_df['_time'] = temp_df['_time'].dt.tz_localize(None)
        temp_df['_time'] = temp_df['_time'].dt.strftime('%Y-%m-%d %H:%M:%S')
        temp_df['_time'] = pd.to_datetime(temp_df['_time'])
        #temp_df['_value'] = temp_df['_value'].clip(lower=0, upper=1)  # Ensure _value is binary (0 or 1)
        df = pd.concat([df, temp_df]).drop_duplicates(subset=['_time']).sort_values(by='_time').reset_index(drop=True)
    return df
```

```python
calefactores_files = ['./climate devices experiments data/calefactores_0.csv', './climate devices experiments data/calefactores_1.csv', 
             './climate devices experiments data/calefactores_2.csv']
df_calefactores = readListCSV(calefactores_files)
df_calefactores
```

```python
recirculador_files = ['./climate devices experiments data/Recirculador_0.csv', './climate devices experiments data/Recirculador_1.csv']
df_recirculador = readListCSV(recirculador_files)
df_recirculador
```
