```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
import datetime
from datetime import timedelta

# calculate CVRMSE
def calculate_cvrmse(y_true, y_pred):
    rmse = root_mean_squared_error(y_true, y_pred)
    cvrmse = rmse / y_true.mean()
    return cvrmse
```

```python
def showMetrics(forecast_df, exterior_df, variable_name):
    # join the two dataframes on the date column
    forecast_df = forecast_df.set_index('date')
    exterior_df = exterior_df.set_index('date')
    joined_df = forecast_df.join(exterior_df.dropna(), how='inner')

    mae = mean_absolute_error(joined_df['_value'], joined_df[variable_name])
    cvrmse = calculate_cvrmse(joined_df['_value'], joined_df[variable_name])
    print(f'{variable_name}: MAE: {mae}, CVRMSE: {cvrmse}')
    return joined_df

# calculate MAE and CVRMSE for each pair variable between forecast and exterior data
joined_temp =showMetrics(df_forecast[['date', 'temperature_2m']], df_temperatura_exterior.rename(columns={'_time': 'date'}), 'temperature_2m')
showMetrics(df_forecast[['date', 'relative_humidity_2m']], df_humedad_exterior.rename(columns={'_time': 'date'}), 'relative_humidity_2m')
showMetrics(df_forecast[['date', 'wind_speed_10m']], df_velocidad_viento_exterior.rename(columns={'_time': 'date'}), 'wind_speed_10m')
showMetrics(df_forecast[['date', 'precipitation']], df_precipitacion_exterior.rename(columns={'_time': 'date'}), 'precipitation')
print()
```

# Climatic control system variables

```python
# read multiple CSV files and combine them into a single DataFrame
def readListCSV(csv_files):
    df = pd.DataFrame()
    for file in csv_files:
        temp_df = pd.read_csv(file)
        temp_df['_time'] = pd.to_datetime(temp_df['_time'], format='mixed', utc=True)
        # remove the timezone information from the _time column
        temp_df['_time'] = temp_df['_time'].dt.tz_localize(None)
        temp_df['_time'] = temp_df['_time'].dt.strftime('%Y-%m-%d %H:%M:%S')
        temp_df['_time'] = pd.to_datetime(temp_df['_time'])
        temp_df['_value'] = temp_df['_value'].clip(lower=0, upper=1)  # Ensure _value is binary (0 or 1)
        df = pd.concat([df, temp_df]).drop_duplicates(subset=['_time']).sort_values(by='_time').reset_index(drop=True)
    return df

# calculate the hourly duration of the active state (ON) 
def calculate_hourly_duration(df, active_value):
    df = df.copy().set_index('_time')
    '''
    if '_time' in df.columns:
        df['_time'] = pd.to_datetime(df['_time'], utc=True, errors='coerce').dt.tz_localize(None)
        df = df.sort_values('_time').set_index('_time')
    else:
        df.index = pd.to_datetime(df.index, utc=True, errors='coerce')
        if getattr(df.index, 'tz', None) is not None:
            df.index = df.index.tz_localize(None)
        df = df.sort_index()
    '''
    # Each row represents the state from its timestamp until the next timestamp.
    intervals = df.reset_index()
    intervals = intervals.rename(columns={intervals.columns[0]: '_time'})
    intervals['next_time'] = intervals['_time'].shift(-1)
    intervals['duration_hours'] = (intervals['next_time'] - intervals['_time']).dt.total_seconds() / 3600

    # Split each ON interval across hourly buckets and sum it.
    rows = []
    for _, row in intervals.dropna(subset=['next_time']).iterrows():
        if row['_value'] == active_value:
            start = row['_time']
            end = row['next_time']
            current = start.floor('h')
            while current < end:
                next_hour = current + pd.Timedelta(hours=1)
                overlap = min(end, next_hour) - max(start, current)
                if overlap.total_seconds() > 0:
                    rows.append((current, overlap.total_seconds() / 60))
                current = next_hour

    hourly_on_minutes = (
        pd.DataFrame(rows, columns=['_time', 'on_minutes'])
        .groupby('_time', as_index=False)['on_minutes']
        .sum()
    )

    return hourly_on_minutes

# read multiple CSV files and combine them into a single DataFrame
def readListCSV_windows_pantalla(csv_files):
    df = pd.DataFrame()
    for file in csv_files:
        temp_df = pd.read_csv(file)
        temp_df['_time'] = pd.to_datetime(temp_df['_time'], format='mixed', utc=True)
        # remove the timezone information from the _time column
        temp_df['_time'] = temp_df['_time'].dt.tz_localize(None)
        temp_df['_time'] = temp_df['_time'].dt.strftime('%Y-%m-%d %H:%M:%S')
        temp_df['_time'] = pd.to_datetime(temp_df['_time'])
        temp_df['_value'] = temp_df['_value'].clip(lower=0)  # Ensure _value is binary (0 or 1)
        df = pd.concat([df, temp_df]).drop_duplicates(subset=['_time']).sort_values(by='_time').reset_index(drop=True)
    return df


# calculate the hourly duration of the active state (ON) 
def calculate_hourly_duration_windows_pantalla(df, active_value):
    df = df.copy().set_index('_time')
    '''
    if '_time' in df.columns:
        df['_time'] = pd.to_datetime(df['_time'], utc=True, errors='coerce').dt.tz_localize(None)
        df = df.sort_values('_time').set_index('_time')
    else:
        df.index = pd.to_datetime(df.index, utc=True, errors='coerce')
        if getattr(df.index, 'tz', None) is not None:
            df.index = df.index.tz_localize(None)
        df = df.sort_index()
    '''
    # Each row represents the state from its timestamp until the next timestamp.
    intervals = df.reset_index()
    intervals = intervals.rename(columns={intervals.columns[0]: '_time'})
    intervals['next_time'] = intervals['_time'].shift(-1)
    intervals['duration_hours'] = (intervals['next_time'] - intervals['_time']).dt.total_seconds() / 3600

    # Split each ON interval across hourly buckets and sum it.
    rows = []
    for _, row in intervals.dropna(subset=['next_time']).iterrows():
        if row['_value'] > active_value:
            start = row['_time']
            end = row['next_time']
            current = start.floor('h')
            while current < end:
                next_hour = current + pd.Timedelta(hours=1)
                overlap = min(end, next_hour) - max(start, current)
                if overlap.total_seconds() > 0:
                    rows.append((current, overlap.total_seconds() / 60))
                current = next_hour

    hourly_on_minutes = (
        pd.DataFrame(rows, columns=['_time', 'on_minutes'])
        .groupby('_time', as_index=False)['on_minutes']
        .sum()
    )

    return hourly_on_minutes
```
